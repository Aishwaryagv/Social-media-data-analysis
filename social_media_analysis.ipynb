```python
import pandas as pd

df = pd.read_csv("Project 3_dataset_answers1.csv")
```

```python
data = df[df.columns[0]].str.split("_", expand=True)
```

```python
data.columns = [
    "qid",
    "i",
    "qs",
    "qt",
    "q_creation_time",
    "tags",
    "qvc",
    "qac",
    "aid",
    "j",
    "ans_score",
    "at"
]
```

```python
data.head()
```

```python
data.info()
```

```python
from collections import Counter

# Split all tags into individual tags
all_tags = data["tags"].str.split(",")

# Flatten the list
flat_tags = [tag.strip() for tags in all_tags for tag in tags]

# Count tags
tag_counts = Counter(flat_tags)

# Display Top 10
top10 = pd.DataFrame(tag_counts.most_common(10), columns=["Tag", "Count"])

top10
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.bar(top10["Tag"], top10["Count"])
plt.title("Top 10 Most Used Tags")
plt.xlabel("Tags")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Convert timestamps to numbers
data["q_creation_time"] = pd.to_numeric(data["q_creation_time"])
data["at"] = pd.to_numeric(data["at"])

# Calculate answer time in seconds
data["answer_time_sec"] = data["at"] - data["q_creation_time"]

# Convert to minutes
data["answer_time_min"] = data["answer_time_sec"] / 60

# Display average
avg_time = data["answer_time_min"].mean()

print("Average Time to Answer (minutes):", avg_time)
```

```python
avg_minutes = data["answer_time_min"].mean()

avg_hours = avg_minutes / 60
avg_days = avg_hours / 24

print(f"Average Time to Answer: {avg_minutes:.2f} minutes")
print(f"Average Time to Answer: {avg_hours:.2f} hours")
print(f"Average Time to Answer: {avg_days:.2f} days")
```

```python
# Calculate time difference in seconds
data["answer_time_sec"] = data["at"] - data["q_creation_time"]

# Filter questions answered within 1 hour (3600 seconds)
within_one_hour = data[data["answer_time_sec"] <= 3600]

# Display the count
print("Number of questions answered within 1 hour:", len(within_one_hour))
```

```python
from collections import Counter

# Split tags from questions answered within 1 hour
quick_tags = within_one_hour["tags"].str.split(",")

# Flatten into one list
flat_tags = [tag.strip() for tags in quick_tags for tag in tags]

# Count occurrences
quick_tag_counts = Counter(flat_tags)

# Convert to DataFrame
quick_tag_df = pd.DataFrame(
    quick_tag_counts.items(),
    columns=["Tag", "Count"]
)

# Sort by frequency
quick_tag_df = quick_tag_df.sort_values(
    by="Count",
    ascending=False
)

quick_tag_df.head(20)
```

```python
import matplotlib.pyplot as plt

top10 = quick_tag_df.head(10)

plt.figure(figsize=(10,5))
plt.bar(top10["Tag"], top10["Count"])
plt.title("Top Tags Answered Within 1 Hour")
plt.xlabel("Tags")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```
